```python
import os
import json
import cv2
import numpy as np
from scipy.ndimage import label, center_of_mass
from tqdm.notebook import tqdm

def get_lesion_info(mask_path, optic_disc_center=None):
    mask = cv2.imread(mask_path, cv2.IMREAD_GRAYSCALE)
    if mask is None or np.max(mask) == 0:
        return {"count": 0, "avg_size": 0, "avg_distance_to_disc": None}
    
    labeled_array, num_features = label(mask > 0)
    sizes = [(labeled_array == i).sum() for i in range(1, num_features + 1)]
    avg_size = np.mean(sizes) if sizes else 0

    if optic_disc_center:
        coords = center_of_mass(mask, labeled_array, range(1, num_features + 1))
        distances = [np.linalg.norm(np.array(c) - np.array(optic_disc_center)) for c in coords]
        avg_distance = np.mean(distances) if distances else None
    else:
        avg_distance = None

    return {
        "count": num_features,
        "avg_size": avg_size,
        "avg_distance_to_disc": avg_distance
    }

base_dir = "/Users/poojasrisundaresan/Desktop/Internship/A. Segmentation/2. All Segmentation Groundtruths/a. Training Set"

ma_dir = os.path.join(base_dir, "1. Microaneurysms")
he_dir = os.path.join(base_dir, "2. Hemorrhages")
ex_dir = os.path.join(base_dir, "3. Hard Exudates")
se_dir = os.path.join(base_dir, "4. Soft Exudates")
od_dir = os.path.join(base_dir, "5. Optic Disc")

image_ids = sorted([f.split('.')[0] for f in os.listdir(ma_dir) if f.endswith('.tif')])
```

```python
optic_disc_centers = {}

for image_id in tqdm(image_ids, desc="Processing Optic Disc"):
    mask_path = os.path.join(od_dir, f"{image_id}.tif")
    mask = cv2.imread(mask_path, cv2.IMREAD_GRAYSCALE)
    if mask is not None and np.max(mask) > 0:
        labeled_array, _ = label(mask > 0)
        centers = center_of_mass(mask, labeled_array, [1])
        optic_disc_centers[image_id] = centers[0]
    else:
        optic_disc_centers[image_id] = None
```

```python
ma_data = {}
for image_id in tqdm(image_ids, desc="Processing MA"):
    mask_path = os.path.join(ma_dir, f"{image_id}.tif")
    ma_data[image_id] = get_lesion_info(mask_path, optic_disc_centers[image_id])
```

```python
he_data = {}
for image_id in tqdm(image_ids, desc="Processing HE"):
    mask_path = os.path.join(he_dir, f"{image_id}.tif")
    he_data[image_id] = get_lesion_info(mask_path, optic_disc_centers[image_id])
```

```python
ex_data = {}
for image_id in tqdm(image_ids, desc="Processing EX"):
    mask_path = os.path.join(ex_dir, f"{image_id}.tif")
    ex_data[image_id] = get_lesion_info(mask_path, optic_disc_centers[image_id])
```

```python
se_data = {}
for image_id in tqdm(image_ids, desc="Processing SE"):
    mask_path = os.path.join(se_dir, f"{image_id}.tif")
    se_data[image_id] = get_lesion_info(mask_path, optic_disc_centers[image_id])
```

```python
def generate_report(ma, he, ex, se):
    sentences = []

    if ma["count"] > 0:
        sentences.append("Few microaneurysms detected." if ma["count"] < 5 else f"{ma['count']} microaneurysms detected.")
    else:
        sentences.append("No microaneurysms observed.")

    if he["count"] > 0:
        sentences.append(f"{he['count']} retinal hemorrhages present.")
    else:
        sentences.append("No hemorrhages observed.")

    if ex["count"] > 0:
        location = "near the optic disc" if ex["avg_distance_to_disc"] and ex["avg_distance_to_disc"] < 100 else "in the peripheral retina"
        sentences.append(f"Hard exudates located {location}.")
    else:
        sentences.append("No hard exudates detected.")

    if se["count"] > 0:
        sentences.append("Soft exudates detected.")
    else:
        sentences.append("No soft exudates seen.")

    return " ".join(sentences)

# Final report generation
synthetic_reports = []

for image_id in image_ids:
    report = generate_report(ma_data[image_id], he_data[image_id], ex_data[image_id], se_data[image_id])
    result = {
        "image_id": image_id,
        "report": report,
        "labels": {
            "microaneurysms": ma_data[image_id]["count"] > 0,
            "hemorrhages": he_data[image_id]["count"] > 0,
            "hard_exudates": ex_data[image_id]["count"] > 0,
            "soft_exudates": se_data[image_id]["count"] > 0
        }
    }
    synthetic_reports.append(result)

# Display a few samples
for r in synthetic_reports[:5]:
    print(f"Image ID: {r['image_id']}")
    print("Report:", r['report'])
    print("Labels:", r['labels'])
    print("-" * 50)

# Save to disk
with open("synthetic_reports.json", "w") as f:
    json.dump(synthetic_reports, f, indent=4)
```
